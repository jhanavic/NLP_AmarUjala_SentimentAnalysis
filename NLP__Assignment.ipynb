```python
# 1. Install the specialized Hindi NLP library
!pip install stanza

import stanza
import os
import pandas as pd # Import pandas

# 2. Download and Load the Hindi NER model
print("Downloading Hindi NLP Model...")
stanza.download('hi')
nlp = stanza.Pipeline('hi', processors='tokenize,ner')

# 3. Define the extraction function
def get_model_entities(text):
    if not isinstance(text, str) or text.strip() == "":
        return "None"

    # Run the AI model on the text
    doc = nlp(text)

    # Extract the identified entities
    entities = []
    for ent in doc.ents:
        entities.append(f"{ent.text} ({ent.type})")

    return ", ".join(list(set(entities))) if entities else "No Entities Found"

# 4. Load your CSV
files = [f for f in os.listdir('.') if 'Final_Cleaned_Dataset' in f and f.endswith('.csv')]
if files:
    df = pd.read_csv(files[0])

    print(f"Processing {len(df)} rows. This may take 1-2 minutes...")

    # 5. Apply to the entire dataset
    df['AI_Named_Entities'] = df['Text_No_Stopwords'].apply(get_model_entities)

    # 6. Final Output File
    output_file_csv = 'AI_Hindi_NER_Final_Report.csv'
    output_file_excel = 'AI_Hindi_NER_Final_Report.xlsx'

    df.to_csv(output_file_csv, index=False)
    df.to_excel(output_file_excel, index=False)

    print(f"Success! Full report saved to: {output_file_csv} and {output_file_excel}")

    # Download the files automatically
    from google.colab import files as colab_files
    colab_files.download(output_file_csv)
    colab_files.download(output_file_excel)
else:
    print("Error: Please ensure the 'Final_Cleaned_Dataset' CSV is uploaded to your Colab files.")
```

### 2. Part-of-Speech (POS) Tagging
For POS tagging, libraries like `Stanza` (formerly StanfordNLP) offer robust models for Hindi. This allows you to identify the grammatical role of each word (e.g., noun, verb, adjective).

```python
# 1. Download and Load the Hindi POS model
print("Downloading Hindi POS Tagger...")
stanza.download('hi')
nlp = stanza.Pipeline('hi', processors='tokenize,pos')

# 2. Load the output from the NER step
try:
    df = pd.read_csv('AI_Hindi_NER_Final_Report.csv')
    print("NER file loaded successfully.")
except FileNotFoundError:
    print("Error: 'AI_Hindi_NER_Final_Report.csv' not found. Please upload it or check the filename.")

# 3. Define the POS extraction function
def get_pos_tags(text):
    if not isinstance(text, str) or text.strip() == "":
        return "None"

    # Process text with the AI model
    doc = nlp(text)

    # Extract word and its Universal POS tag (UPOS)
    pos_list = []
    for sent in doc.sentences:
        for word in sent.words:
            pos_list.append(f"{word.text} ({word.upos})")

    return ", ".join(pos_list)

# 4. Apply POS Tagging to the entire dataset
print("Starting POS tagging for all rows. This may take a moment...")
df['POS_Tags'] = df['Text_No_Stopwords'].apply(get_pos_tags)

# 5. Save the final multi-step analysis
output_csv = 'Hindi_News_Full_NLP_Report.csv'
output_excel = 'Hindi_News_Full_NLP_Report.xlsx'

df.to_csv(output_csv, index=False)
df.to_excel(output_excel, index=False)

print(f"Success! Final report saved as {output_excel}")

#Download the file automatically
from google.colab import files
files.download(output_csv)
files.download(output_excel)
```

n-grams

```python
from nltk.util import ngrams

def generate_bigrams(text):

    tokens = str(text).split()

    bigrams = list(
        ngrams(tokens, 2)
    )

    return ", ".join(
        [" ".join(bg) for bg in bigrams]
    )

def generate_trigrams(text):

    tokens = str(text).split()

    trigrams = list(
        ngrams(tokens, 3)
    )

    return ", ".join(
        [" ".join(tg) for tg in trigrams]
    )

df["Bigrams"] = df[
    "Text_No_Stopwords"
].apply(generate_bigrams)

df["Trigrams"] = df[
    "Text_No_Stopwords"
].apply(generate_trigrams)

output_csv = "Hindi_News_With_Ngrams.csv"
output_excel = "Hindi_News_With_Ngrams.xlsx"
df.to_csv(
    output_csv,
    index=False
)
df.to_excel(
    output_excel,
    index=False
)
from google.colab import files
files.download(output_csv)
files.download(output_excel)
```

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    classification_report,
    accuracy_score
)

# =========================================================
# LOAD DATASET
# =========================================================

try:
    df = pd.read_csv(
        "Hindi_News_Labelled_Dataset.csv" # Corrected filename
    )

    print("Dataset Loaded Successfully")

except FileNotFoundError:

    print(
        "Error: Hindi_News_Labelled_Dataset.csv not found" # Corrected filename in error message
    )

    exit()

# =========================================================
# COMBINED FEATURES
# =========================================================

df["combined_features"] = (

    df["POS_Tags"].astype(str)

    + " "

    + df["AI_Named_Entities"].astype(str)

    + " "

    + df["Headline_No_Stopwords"].astype(str)

    + " "

    + df["Bigrams"].astype(str)

    + " "

    + df["Trigrams"].astype(str)
)

# =========================================================
# SENTIMENT LABEL ENCODING
# =========================================================

sentiment_encoder = LabelEncoder()

y_sentiment = sentiment_encoder.fit_transform(
    df["Sentiment"]
)

sentiment_classes = (
    sentiment_encoder.classes_
)

# =========================================================
# EMOTION LABEL ENCODING
# =========================================================

emotion_encoder = LabelEncoder()

y_emotion = emotion_encoder.fit_transform(
    df["Emotion"]
)

emotion_classes = (
    emotion_encoder.classes_
)

# =========================================================
# TF-IDF VECTORIZATION
# =========================================================

vectorizer = TfidfVectorizer(

    max_features=3000,

    ngram_range=(1, 3)
)

X = vectorizer.fit_transform(
    df["combined_features"]
)

# =========================================================
# SENTIMENT TRAIN / TEST
# =========================================================

(
    X_train_s,
    X_test_s,
    y_train_s,
    y_test_s

) = train_test_split(

    X,

    y_sentiment,

    train_size=0.8,

    random_state=42,

    stratify=y_sentiment
)

# =========================================================
# EMOTION TRAIN / TEST
# =========================================================

(
    X_train_e,
    X_test_e,
    y_train_e,
    y_test_e

) = train_test_split(

    X,

    y_emotion,

    train_size=0.8,

    random_state=42,

    stratify=y_emotion
)

# =========================================================
# SENTIMENT MODEL
# =========================================================

sentiment_model = RandomForestClassifier(

    n_estimators=300,

    max_depth=None,

    random_state=42,

    class_weight="balanced"
)

sentiment_model.fit(
    X_train_s,
    y_train_s
)

# =========================================================
# EMOTION MODEL
# =========================================================

emotion_model = RandomForestClassifier(

    n_estimators=300,

    max_depth=None,

    random_state=42,

    class_weight="balanced"
)

emotion_model.fit(
    X_train_e,
    y_train_e
)

print("\nTraining Completed")

# =========================================================
# SENTIMENT EVALUATION
# =========================================================

sentiment_pred = sentiment_model.predict(
    X_test_s
)

sentiment_accuracy = accuracy_score(
    y_test_s,
    sentiment_pred
)

print("\n===================================")
print(
    f"SENTIMENT ACCURACY : {sentiment_accuracy*100:.2f}%"
)
print("===================================")

print(

    classification_report(

        y_test_s,

        sentiment_pred,

        target_names=sentiment_classes
    )
)

# =========================================================
# EMOTION EVALUATION
# =========================================================

emotion_pred = emotion_model.predict(
    X_test_e
)

emotion_accuracy = accuracy_score(
    y_test_e,
    emotion_pred
)

print("\n===================================")
print(
    f"EMOTION ACCURACY : {emotion_accuracy*100:.2f}%"
)
print("===================================")

# Get unique labels present in the test set
unique_test_labels_e = np.unique(y_test_e)

# Get corresponding target names
target_names_for_report_e = (
    emotion_encoder.inverse_transform(
        unique_test_labels_e
    )
)

print(

    classification_report(

        y_test_e,

        emotion_pred,

        labels=unique_test_labels_e,

        target_names=target_names_for_report_e
    )
)

# =========================================================
# PREDICT FOR ALL ROWS
# =========================================================

all_sentiment_preds = sentiment_model.predict(
    X
)

all_emotion_preds = emotion_model.predict(
    X
)

df["Predicted_Sentiment"] = (

    sentiment_encoder.inverse_transform(
        all_sentiment_preds
    )
)

df["Predicted_Emotion"] = (

    emotion_encoder.inverse_transform(
        all_emotion_preds
    )
)

# =========================================================
# EXPORT FINAL REPORT
# =========================================================

output_file = (
    "Hindi_News_Sentiment_Emotion_Report.csv"
)

df.to_csv(

    output_file,

    index=False,

    encoding="utf-8-sig"
)

print(
    f"\nReport Saved Successfully : {output_file}"
)

# =========================================================
# DOWNLOAD FILE
# =========================================================

from google.colab import files

files.download(
    output_file
)
```

```python
# =========================================================
# INSTALL LIBRARIES
# =========================================================
# !pip install -q transformers datasets accelerate gradio scikit-learn pandas torch

# =========================================================
# IMPORT LIBRARIES
# =========================================================
import gc
import pandas as pd
import torch
import torch.nn as nn
import numpy as np
import gradio as gr

from datasets import Dataset
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import classification_report, accuracy_score, f1_score

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    TrainingArguments,
    Trainer
)

# Set up device safely
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# =========================================================
# LOAD DATASET
# =========================================================
df = pd.read_csv("Hindi_News_Labelled_Dataset.csv")
df = df.dropna()

# Clean and normalize category labels
df["Sentiment"] = df["Sentiment"].astype(str).str.strip().str.capitalize()
df["Emotion"] = df["Emotion"].astype(str).str.strip().str.capitalize()

# Explicit filter syntax to exclude missing entries
df = df[(df["Sentiment"] != "") & (df["Emotion"] != "")]

print("Dataset Shape :", df.shape)
print("\nSentiment Distribution:\n", df["Sentiment"].value_counts())
print("\nEmotion Distribution:\n", df["Emotion"].value_counts())

# =========================================================
# CREATE INPUT TEXT
# =========================================================
df["text"] = (
    df["Headline_No_Stopwords"].astype(str)
    + " "
    + df["Text_No_Stopwords"].astype(str)
)

# =========================================================
# LABEL ENCODERS
# =========================================================
sentiment_encoder = LabelEncoder()
emotion_encoder = LabelEncoder()

df["sentiment_label"] = sentiment_encoder.fit_transform(df["Sentiment"])
df["emotion_label"] = emotion_encoder.fit_transform(df["Emotion"])

# =========================================================
# CUSTOM FOCAL WEIGHTED TRAINER CLASS
# =========================================================
class BalancedTrainer(Trainer):
    def __init__(self, class_weights=None, **kwargs):
        super().__init__(**kwargs)
        if class_weights is not None:
            self.class_weights = torch.tensor(class_weights, dtype=torch.float).to(device)
        else:
            self.class_weights = None

    def compute_loss(self, model, inputs, return_outputs=False, num_items_in_batch=None):
        labels = inputs.get("labels")
        outputs = model(**inputs)
        logits = outputs.get("logits")

        # Cross entropy with penalization weights to scale up minority classes like Positive
        if self.class_weights is not None:
            loss_fct = nn.CrossEntropyLoss(weight=self.class_weights)
        else:
            loss_fct = nn.CrossEntropyLoss()

        loss = loss_fct(logits.view(-1, self.model.config.num_labels), labels.view(-1))
        return (loss, outputs) if return_outputs else loss

# Evaluation helper focusing heavily on Macro-F1 performance
def compute_metrics(eval_pred):
    labels = eval_pred.label_ids
    preds = eval_pred.predictions.argmax(-1)
    acc = accuracy_score(labels, preds)
    macro_f1 = f1_score(labels, preds, average="macro")
    return {"accuracy": acc, "macro_f1": macro_f1}

# =========================================================
# MODEL SETUP
# =========================================================
MODEL_NAME = "ai4bharat/IndicBERTv2-MLM-only"
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

# =========================================================
# TRAINING ENGINE
# =========================================================
def train_model(label_column, output_dir):
    class_labels = np.unique(df[label_column])

    # Generate strict balanced class weights
    weights = compute_class_weight(
        class_weight='balanced',
        classes=class_labels,
        y=df[label_column].values
    )
    print(f"\n[INFO] Enforcing balanced weights for {label_column}: {weights}")

    # Stratified split forces minority labels like "Positive" into both pools evenly
    train_df, test_df = train_test_split(
        df,
        test_size=0.2,
        random_state=42,
        stratify=df[label_column]
    )

    train_dataset = Dataset.from_pandas(train_df[["text", label_column]].rename(columns={label_column: "labels"}))
    test_dataset = Dataset.from_pandas(test_df[["text", label_column]].rename(columns={label_column: "labels"}))

    def tokenize_func(batch):
        return tokenizer(batch["text"], truncation=True, padding="max_length", max_length=256)

    train_dataset = train_dataset.map(tokenize_func, batched=True, remove_columns=["text"])
    test_dataset = test_dataset.map(tokenize_func, batched=True, remove_columns=["text"])

    train_dataset.set_format("torch")
    test_dataset.set_format("torch")

    model = AutoModelForSequenceClassification.from_pretrained(
        MODEL_NAME,
        num_labels=len(class_labels)
    )

    # Balanced hyperparameters
    training_args = TrainingArguments(
        output_dir=output_dir,
        learning_rate=2e-5,               # Conservative learning rate to capture minor features carefully
        per_device_train_batch_size=8,
        per_device_eval_batch_size=8,
        num_train_epochs=5,               # Incremented to 5 epochs for slower, thorough adaptation
        weight_decay=0.01,
        logging_steps=10,
        eval_strategy="epoch",
        save_strategy="epoch",
        metric_for_best_model="macro_f1",  # Guarantees the save favors balanced class accuracy over bulk accuracy
        load_best_model_at_end=True,
        save_total_limit=1,
        report_to="none"
    )

    trainer = BalancedTrainer(
        model=model,
        args=training_args,
        train_dataset=train_dataset,
        eval_dataset=test_dataset,
        compute_metrics=compute_metrics,
        class_weights=weights
    )

    print(f"Training {output_dir}...")
    trainer.train()

    trainer.save_model(output_dir)
    tokenizer.save_pretrained(output_dir)

    # Local runtime evaluation sanity check
    predictions = trainer.predict(test_dataset)
    preds = np.argmax(predictions.predictions, axis=-1)
    print(f"\n Validation Performance Report for {label_column}:")
    print(classification_report(predictions.label_ids, preds))

    del model, trainer
    gc.collect()
    torch.cuda.empty_cache()

# =========================================================
# RUN TRAINING PIPELINE
# =========================================================
print("\n============================")
print("TRAINING SENTIMENT MODEL")
print("============================")
train_model("sentiment_label", "sentiment_model")

print("\n============================")
print("TRAINING EMOTION MODEL")
print("============================")
train_model("emotion_label", "emotion_model")

# =========================================================
# LOAD CONFIGURATIONS FOR APP
# =========================================================
sentiment_model = AutoModelForSequenceClassification.from_pretrained("sentiment_model").to(device)
emotion_model = AutoModelForSequenceClassification.from_pretrained("emotion_model").to(device)

sentiment_model.eval()
emotion_model.eval()

# =========================================================
# PREDICTION FUNCTION
# =========================================================
def predict_label(text, model, encoder):
    inputs = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=256)
    inputs = {k: v.to(device) for k, v in inputs.items()}

    with torch.no_grad():
        outputs = model(**inputs)

    probs = torch.softmax(outputs.logits, dim=1)
    pred_id = torch.argmax(probs, dim=1).item()
    confidence = probs[0][pred_id].item()

    label = encoder.inverse_transform([pred_id])[0]
    return label, round(confidence * 100, 2)

# =========================================================
# GRADIO INTERFACE LOGIC
# =========================================================
def analyze_news(headline, content):
    text = str(headline) + " " + str(content)
    sentiment, sent_conf = predict_label(text, sentiment_model, sentiment_encoder)
    emotion, emo_conf = predict_label(text, emotion_model, emotion_encoder)

    return f"""📰 NEWS ANALYSIS

Sentiment : {sentiment}
Confidence : {sent_conf}%

Emotion : {emotion}
Confidence : {emo_conf}%"""

dropdown_choices = df["Headline_No_Stopwords"].tolist()

def load_article(selected_headline):
    row = df[df["Headline_No_Stopwords"] == selected_headline]
    return row.iloc[0]["Headline_No_Stopwords"], row.iloc[0]["Text_No_Stopwords"]

with gr.Blocks(theme=gr.themes.Soft()) as demo:
    gr.Markdown("# 📰 Hindi News Sentiment & Emotion Analyst")
    dropdown = gr.Dropdown(choices=dropdown_choices, label="Select News Article")

    with gr.Row():
        with gr.Column():
            headline_box = gr.Textbox(label="Headline")
            content_box = gr.Textbox(label="News Content", lines=8)
            analyze_btn = gr.Button("Analyze News", variant="primary")
        with gr.Column():
            output_box = gr.Textbox(label="Prediction", lines=8)

    dropdown.change(fn=load_article, inputs=dropdown, outputs=[headline_box, content_box])
    analyze_btn.click(fn=analyze_news, inputs=[headline_box, content_box], outputs=output_box)

demo.launch(debug=True)
```

```python
# =========================================================
# INSTALL LIBRARIES (Run this in your terminal if needed)
# =========================================================
# !pip install -q transformers datasets accelerate gradio scikit-learn pandas torch

# =========================================================
# IMPORT LIBRARIES
# =========================================================
import gc
import pandas as pd
import torch
import torch.nn as nn
import numpy as np
import gradio as gr

from datasets import Dataset
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import classification_report, accuracy_score, f1_score

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    TrainingArguments,
    Trainer
)

# Set up device safely
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# =========================================================
# LOAD DATASET
# =========================================================
df = pd.read_csv("Hindi_News_Labelled_Dataset.csv")
df = df.dropna()

# Clean and normalize category labels
df["Sentiment"] = df["Sentiment"].astype(str).str.strip().str.capitalize()
df["Emotion"] = df["Emotion"].astype(str).str.strip().str.capitalize()

# Explicit filter syntax to exclude missing entries
df = df[(df["Sentiment"] != "") & (df["Emotion"] != "")]

print("Dataset Shape :", df.shape)
print("\nSentiment Distribution:\n", df["Sentiment"].value_counts())
print("\nEmotion Distribution:\n", df["Emotion"].value_counts())

# =========================================================
# CREATE INPUT TEXT
# =========================================================
# Retained text formatting, but the tokenizer will process it smoothly
df["text"] = (
    df["Headline_No_Stopwords"].astype(str)
    + " "
    + df["Text_No_Stopwords"].astype(str)
)

# =========================================================
# LABEL ENCODERS
# =========================================================
sentiment_encoder = LabelEncoder()
emotion_encoder = LabelEncoder()

df["sentiment_label"] = sentiment_encoder.fit_transform(df["Sentiment"])
df["emotion_label"] = emotion_encoder.fit_transform(df["Emotion"])

# =========================================================
# FIXED CUSTOM BALANCED TRAINER CLASS
# =========================================================
class BalancedTrainer(Trainer):
    def __init__(self, class_weights=None, **kwargs):
        super().__init__(**kwargs)
        if class_weights is not None:
            # Keep as a CPU tensor initially; don't force a static device here
            self.class_weights = torch.tensor(class_weights, dtype=torch.float)
        else:
            self.class_weights = None

    def compute_loss(self, model, inputs, return_outputs=False, num_items_in_batch=None):
        labels = inputs.get("labels")
        outputs = model(**inputs)
        logits = outputs.get("logits")

        # FIX: Dynamically map weights to the precise device the current batch/labels are on
        if self.class_weights is not None:
            weights = self.class_weights.to(labels.device)
            loss_fct = nn.CrossEntropyLoss(weight=weights)
        else:
            loss_fct = nn.CrossEntropyLoss()

        loss = loss_fct(logits.view(-1, self.model.config.num_labels), labels.view(-1))
        return (loss, outputs) if return_outputs else loss

# Evaluation helper focusing heavily on Macro-F1 performance
def compute_metrics(eval_pred):
    labels = eval_pred.label_ids
    preds = eval_pred.predictions.argmax(-1)
    acc = accuracy_score(labels, preds)
    macro_f1 = f1_score(labels, preds, average="macro")
    return {"accuracy": acc, "macro_f1": macro_f1}

# =========================================================
# MODEL SETUP
# =========================================================
MODEL_NAME = "ai4bharat/IndicBERTv2-MLM-only"
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

# =========================================================
# TRAINING ENGINE
# =========================================================
def train_model(label_column, output_dir):
    class_labels = np.unique(df[label_column])

    # Generate strict balanced class weights
    weights = compute_class_weight(
        class_weight='balanced',
        classes=class_labels,
        y=df[label_column].values
    )
    print(f"\n[INFO] Enforcing balanced weights for {label_column}: {weights}")

    # Stratified split forces minority labels into both pools evenly
    train_df, test_df = train_test_split(
        df,
        test_size=0.2,
        random_state=42,
        stratify=df[label_column]
    )

    train_dataset = Dataset.from_pandas(train_df[["text", label_column]].rename(columns={label_column: "labels"}))
    test_dataset = Dataset.from_pandas(test_df[["text", label_column]].rename(columns={label_column: "labels"}))

    def tokenize_func(batch):
        return tokenizer(batch["text"], truncation=True, padding="max_length", max_length=256)

    train_dataset = train_dataset.map(tokenize_func, batched=True, remove_columns=["text"])
    test_dataset = test_dataset.map(tokenize_func, batched=True, remove_columns=["text"])

    train_dataset.set_format("torch")
    test_dataset.set_format("torch")

    model = AutoModelForSequenceClassification.from_pretrained(
        MODEL_NAME,
        num_labels=len(class_labels)
    )

    # Optimized Hyperparameters to prevent early model collapse / majority-class cheating
    training_args = TrainingArguments(
        output_dir=output_dir,
        learning_rate=1e-5,               # Dropped slightly to safely stabilize minor class feature detection
        per_device_train_batch_size=16,   # Increased batch size to lower gradient noise in cross-entropy scaling
        per_device_eval_batch_size=16,
        num_train_epochs=3,               # Reduced to 3 to prevent over-fitting / loss of minority representation
        weight_decay=0.02,                # Slightly aggressive regularization
        logging_steps=10,
        eval_strategy="epoch",
        save_strategy="epoch",
        metric_for_best_model="macro_f1",  # Guarantees the save favors balanced class accuracy over bulk accuracy
        load_best_model_at_end=True,
        save_total_limit=1,
        report_to="none"
    )

    trainer = BalancedTrainer(
        model=model,
        args=training_args,
        train_dataset=train_dataset,
        eval_dataset=test_dataset,
        compute_metrics=compute_metrics,
        class_weights=weights
    )

    print(f"Training {output_dir}...")
    trainer.train()

    trainer.save_model(output_dir)
    tokenizer.save_pretrained(output_dir)

    # Local runtime evaluation sanity check
    predictions = trainer.predict(test_dataset)
    preds = np.argmax(predictions.predictions, axis=-1)
    print(f"\n Validation Performance Report for {label_column}:")
    print(classification_report(predictions.label_ids, preds))

    del model, trainer
    gc.collect()
    torch.cuda.empty_cache()

# =========================================================
# RUN TRAINING PIPELINE
# =========================================================
print("\n============================")
print("TRAINING SENTIMENT MODEL")
print("============================")
train_model("sentiment_label", "sentiment_model")

print("\n============================")
print("TRAINING EMOTION MODEL")
print("============================")
train_model("emotion_label", "emotion_model")

# =========================================================
# LOAD CONFIGURATIONS FOR APP
# =========================================================
sentiment_model = AutoModelForSequenceClassification.from_pretrained("sentiment_model").to(device)
emotion_model = AutoModelForSequenceClassification.from_pretrained("emotion_model").to(device)

sentiment_model.eval()
emotion_model.eval()

# =========================================================
# PREDICTION FUNCTION
# =========================================================
def predict_label(text, model, encoder):
    inputs = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=256)
    inputs = {k: v.to(device) for k, v in inputs.items()}

    with torch.no_grad():
        outputs = model(**inputs)

    probs = torch.softmax(outputs.logits, dim=1)
    pred_id = torch.argmax(probs, dim=1).item()
    confidence = probs[0][pred_id].item()

    label = encoder.inverse_transform([pred_id])[0]
    return label, round(confidence * 100, 2)

# =========================================================
# GRADIO INTERFACE LOGIC
# =========================================================
def analyze_news(headline, content):
    text = str(headline) + " " + str(content)
    sentiment, sent_conf = predict_label(text, sentiment_model, sentiment_encoder)
    emotion, emo_conf = predict_label(text, emotion_model, emotion_encoder)

    return f"""📰 NEWS ANALYSIS

Sentiment : {sentiment}
Confidence : {sent_conf}%

Emotion : {emotion}
Confidence : {emo_conf}%"""

dropdown_choices = df["Headline_No_Stopwords"].tolist()

def load_article(selected_headline):
    row = df[df["Headline_No_Stopwords"] == selected_headline]
    return row.iloc[0]["Headline_No_Stopwords"], row.iloc[0]["Text_No_Stopwords"]

with gr.Blocks(theme=gr.themes.Soft()) as demo:
    gr.Markdown("# 📰 Hindi News Sentiment & Emotion Analyst")
    dropdown = gr.Dropdown(choices=dropdown_choices, label="Select News Article")

    with gr.Row():
        with gr.Column():
            headline_box = gr.Textbox(label="Headline")
            content_box = gr.Textbox(label="News Content", lines=8)
            analyze_btn = gr.Button("Analyze News", variant="primary")
        with gr.Column():
            output_box = gr.Textbox(label="Prediction", lines=8)

    dropdown.change(fn=load_article, inputs=dropdown, outputs=[headline_box, content_box])
    analyze_btn.click(fn=analyze_news, inputs=[headline_box, content_box], outputs=output_box)

demo.launch(debug=True)
```
